```python
num=float(input("enter first number"))
num1=float(input("enter second number"))
operation = input("please enter operation +, *, -, /")

if operation == "+":
    result = num + num1
elif operation == "*":
    result  = num *num1
elif operation == "-":
    result  = num-num1
elif operation == "/":
    result  = num /num1

else:
    print("invalid operator")
print(result)
```
